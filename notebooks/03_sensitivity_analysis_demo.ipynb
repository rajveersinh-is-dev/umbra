# 03 - MNAR Sensitivity Grid Analysis & Tipping Points

Instead of trusting a single point estimate, sweep over plausible MNAR departures (delta) to evaluate conclusion robustness.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scripts.build_synthetic_benchmarks import generate_benchmark_battery
from umbra import run_sensitivity_grid, explain_sensitivity
```

## 1. Run Sensitivity Grid on Incomplete Data

```python
battery = generate_benchmark_battery(n_samples=2000, random_state=42)
data = battery['MNAR_MEDIUM'].data_observed
report = run_sensitivity_grid(data, target_column='income')
print(report.summary())
```

## 2. Visualize the Sensitivity Curve

```python
plt.figure(figsize=(10, 5))
grid_df = report.grid_df
plt.plot(grid_df['delta'], grid_df['target_mean'], 'b-o', label='Imputed Mean vs Delta')
plt.axhline(report.mar_baseline_estimate, color='r', linestyle='--', label=f'Naive MAR Baseline ({report.mar_baseline_estimate:.2f})')
plt.axvline(0, color='gray', linestyle=':', alpha=0.5)
plt.xlabel('MNAR Shift Delta (Residual Standard Deviations)')
plt.ylabel('Imputed Mean of Income')
plt.title('Sensitivity Curve: Imputed Mean Across MNAR Assumptions')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
